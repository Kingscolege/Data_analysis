**1. Read the CSV File**


```python
import pandas as pd
import numpy as np


df = pd.read_csv("pandas_dataset.csv")

print(" First 5 data")
print(df.head())
original_shape=df.shape
original_shape

```

```python
df.columns = (df.columns.str.strip().str.lower())

df.columns = (df.columns.str.replace(" ", "_"))
df.head()
```

**3. Drop Unnecessary Rows**

```python
# dropping first empty row
df = df.dropna(how="all").reset_index(drop=True)
df.head()
```

```python

print("Shape after removing unnecessary rows:")
print(df.shape)
```

**4. Drop Columns (if needed)**

Since, all column are giving the meaningful information. I dont want to drop any of them.

**5. Count Missing Values**



```python
print("Missing values:")
print(df.isnull().sum())
```

**6. Handle Missing Values**
- Drop rows where `sales` is missing.
- Fill missing values in `profit` using mean and `discount` using median.

```python
sale_drop=df.dropna(subset=['sales'])
df=sale_drop.reset_index(drop=True)
df.head()
```

```python
print(df.isnull().sum())
```

```python
profit_mean = df["profit"].mean()

discount_median = df["discount"].median()

df["profit"] = df["profit"].fillna(profit_mean)

df["discount"] = df["discount"].fillna(discount_median)

df.head()
```

```python
print(df.isnull().sum())
```

**7. Handle Missing Categorical Data**
- Replace missing `customer_name` values with "Unknown".


```python
df["customer_name"] = df["customer_name"].fillna("Unknown")
df.head()
print(df.isnull().sum())
```

**8. Detect and Remove Duplicates**
- Identify duplicate rows based on `order_id` using `df.duplicated()`.
- Remove duplicate rows and verify how many were removed.


```python
duplicate_values = df.duplicated(subset=["order_id"])
duplicate_values
```

```python
duplicate_order_ids = df["order_id"].duplicated()
duplicate_order_ids
```

```python
print("Duplicated Order IDs:")
print(duplicate_order_ids)

print("\nNumber of duplicated Order IDs:")
print(duplicate_order_ids.sum())
```

```python
print("\nDuplicate rows:")
print(df[df["order_id"].duplicated(keep=False)])

# Count rows before removing duplicates
before = len(df)
```

```python
# Remove duplicate Order IDs
df = df.drop_duplicates(
    subset=["order_id"],
    keep="first"
)

# Count rows after removing duplicates
after = len(df)

# Calculate number of rows removed

removed = before - after
print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Number of duplicates removed:", removed)


# Reset index
df = df.reset_index(drop=True)
```

**9. Filtering and Creating New Columns**


```python
print("Orders where unit_price > 20,000:")
print(df[df["unit_price"] > 20000])

print("Orders where unit_price > 10,000 and status is Completed:")
print(df[(df["unit_price"] > 10000) & (df["status"] == "Completed")][["customer_name", "category", "status"]])

df["total_amount"] = df["quantity"] * df["unit_price"] - df["discount"]
print("Dataset with total_amount:")
print(df[["quantity", "unit_price", "discount", "total_amount"]])

df["customer_type"] = np.where(df["quantity"] >= 3, "Bulk Buyer", "Regular Buyer")
print("Dataset with customer_type:")
print(df[["quantity", "customer_type"]])
```
