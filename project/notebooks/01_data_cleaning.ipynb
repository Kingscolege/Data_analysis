# Clean the customer dataset

```python
customers.columns = (
    customers.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
customers.head()
```

```python
baskets.columns = (
    baskets.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
baskets.head()
```

# Correct customer ages

```python
# Convert age to a numeric type.
customers["customer_age"] = pd.to_numeric(
    customers["customer_age"],
)

# Record the original age values before cleaning.
customers["customer_age_original"] = (
    customers["customer_age"]
)
```

```python
# Count ages outside our chosen validation range.
invalid_age_mask = (
    (customers["customer_age"] < 0)
    | (customers["customer_age"] > 100)
)

print("Invalid age records:", invalid_age_mask.sum())

print("\nInvalid age values:")
print(
    customers.loc[
        invalid_age_mask,
        "customer_age"
    ].value_counts().sort_index()
)
```

```python
# Preserve the original values for auditing.
customers["customer_age_original"] = customers["customer_age"].copy()

# Convert the age column to numeric.a
customers["customer_age"] = pd.to_numeric(
    customers["customer_age"],
    errors="coerce"
)

# Identify ages outside the validation range.
invalid_age_mask = ~customers["customer_age"].between(0, 100)

# Report the number of invalid values before modifying them.
print("Invalid ages before cleaning:", invalid_age_mask.sum())

# Mark invalid ages as missing instead of guessing their correct values.
customers.loc[invalid_age_mask, "customer_age"] = np.nan

# Verify the result.
print("\nMissing ages after cleaning:")
print(customers["customer_age"].isna().sum())

print("\nCleaned age statistics:")
print(customers["customer_age"].describe())
```

# Clean the basket dataset

```python
baskets["basket_date"] = pd.to_datetime(
    baskets["basket_date"])

print("Invalid dates:", baskets["basket_date"].isna().sum())
```

# Validate numeric columns

```python
baskets["customer_id"] = pd.to_numeric(
    baskets["customer_id"],
    errors="coerce"
).astype("Int64")
```

```python
baskets["product_id"] = pd.to_numeric(
    baskets["product_id"],
    errors="coerce"
).astype("Int64")

baskets["basket_count"] = pd.to_numeric(
    baskets["basket_count"],
    errors="coerce"
).astype("Int64")
```

# Investigate the customer ID mismatch

```python
customer_ids = set(customers_raw["customer_id"])
basket_customer_ids = set(baskets_raw["customer_id"])

matched_ids = customer_ids & basket_customer_ids
unmatched_basket_ids = basket_customer_ids - customer_ids

print("Unique customers:", len(customer_ids))
print("Unique basket customers:", len(basket_customer_ids))
print("Matching customer IDs:", len(matched_ids))
print("Unmatched basket customer IDs:", len(unmatched_basket_ids))

print(
    "Basket rows without a matching customer:",
    (~baskets_raw["customer_id"].isin(customer_ids)).sum()
)
```

# Validate and export the cleaned data

```python
# Customer validation
assert customers["customer_id"].notna().all()
assert customers["customer_id"].is_unique

# Age validation
assert (
    customers["customer_age"].dropna().between(0, 100).all()
)

# Basket validation
assert baskets["customer_id"].notna().all()
assert baskets["product_id"].notna().all()
assert baskets["basket_date"].notna().all()

assert (
    baskets["basket_count"].dropna().gt(0).all()
)

print("Basic validation passed.")
```

```python
customers_final = customers.drop(
    columns=["customer_age_original"],
)

customers_final.to_csv(
    PROCESSED_DIR / "customers_cleaned.csv",
    index=False
)

baskets.to_csv(
    PROCESSED_DIR / "baskets_cleaned.csv",
    index=False
)

print("Cleaned datasets exported successfully.")
```
